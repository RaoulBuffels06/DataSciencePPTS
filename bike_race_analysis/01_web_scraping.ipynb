```python
# Import required libraries
import requests
from bs4 import BeautifulSoup
import json
import csv
from pathlib import Path
import time
```

## Setup Data Directories

We'll create directories to store our raw scraped data.

```python
# Create data directories if they don't exist
data_dir = Path('data/raw')
data_dir.mkdir(parents=True, exist_ok=True)
print(f"Data directory ready: {data_dir.absolute()}")
```

## Part 1: Scraping Wikipedia for Bicycle Race Information

We'll start by scraping general information about major bicycle races from Wikipedia.

```python
def scrape_wikipedia_races():
    """
    Scrape bicycle race information from Wikipedia.
    Returns a list of dictionaries containing race information.
    If scraping fails, returns sample data for demonstration.
    """
    url = "https://en.wikipedia.org/wiki/List_of_men%27s_road_bicycle_races"
    
    try:
        # Add headers to avoid being blocked
        headers = {
            'User-Agent': 'Mozilla/5.0 (Educational Project)'
        }
        response = requests.get(url, headers=headers, timeout=10)
        response.raise_for_status()
        
        soup = BeautifulSoup(response.content, 'html.parser')
        races = []
        
        # Find tables containing race information
        tables = soup.find_all('table', class_='wikitable')
        
        for table in tables[:3]:  # Process first 3 tables (Grand Tours, Monuments, etc.)
            rows = table.find_all('tr')[1:]  # Skip header row
            
            for row in rows:
                cells = row.find_all(['td', 'th'])
                if len(cells) >= 2:
                    race_name = cells[0].get_text(strip=True)
                    if len(cells) >= 3:
                        country = cells[1].get_text(strip=True)
                        race_type = cells[2].get_text(strip=True) if len(cells) > 2 else 'N/A'
                    else:
                        country = 'N/A'
                        race_type = 'N/A'
                    
                    races.append({
                        'name': race_name,
                        'country': country,
                        'type': race_type
                    })
        
        if races:
            print(f"Successfully scraped {len(races)} races from Wikipedia")
            return races
        else:
            raise ValueError("No races found in tables")
    
    except Exception as e:
        print(f"Note: Could not scrape live Wikipedia data: {e}")
        print("Using sample bicycle race data for demonstration...")
        
        # Return sample data representing major bicycle races
        return [
            {'name': 'Tour de France', 'country': 'France', 'type': 'Grand Tour'},
            {'name': 'Giro d\'Italia', 'country': 'Italy', 'type': 'Grand Tour'},
            {'name': 'Vuelta a España', 'country': 'Spain', 'type': 'Grand Tour'},
            {'name': 'Paris–Roubaix', 'country': 'France', 'type': 'Monument'},
            {'name': 'Milan–San Remo', 'country': 'Italy', 'type': 'Monument'},
            {'name': 'Tour of Flanders', 'country': 'Belgium', 'type': 'Monument'},
            {'name': 'Liège–Bastogne–Liège', 'country': 'Belgium', 'type': 'Monument'},
            {'name': 'Il Lombardia', 'country': 'Italy', 'type': 'Monument'},
            {'name': 'Paris–Nice', 'country': 'France', 'type': 'Stage Race'},
            {'name': 'Tirreno–Adriatico', 'country': 'Italy', 'type': 'Stage Race'},
        ]
```

```python
# Scrape Wikipedia race data
wiki_races = scrape_wikipedia_races()
print(f"\nSample data (first 5 races):")
for race in wiki_races[:5]:
    print(race)
```

## Part 3: Save Scraped Data

Now we'll save our scraped data to both CSV and JSON formats for later processing.

```python
def save_to_csv(data, filename, fieldnames=None):
    """
    Save data to a CSV file.
    """
    if not data:
        print(f"No data to save to {filename}")
        return
    
    filepath = data_dir / filename
    
    if fieldnames is None:
        fieldnames = data[0].keys()
    
    with open(filepath, 'w', newline='', encoding='utf-8') as csvfile:
        writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(data)
    
    print(f"Saved {len(data)} records to {filepath}")

def save_to_json(data, filename):
    """
    Save data to a JSON file.
    """
    if not data:
        print(f"No data to save to {filename}")
        return
    
    filepath = data_dir / filename
    
    with open(filepath, 'w', encoding='utf-8') as jsonfile:
        json.dump(data, jsonfile, indent=2, ensure_ascii=False)
    
    print(f"Saved {len(data)} records to {filepath}")
```

```python
# Save Wikipedia race data
save_to_csv(wiki_races, 'wikipedia_races.csv')
save_to_json(wiki_races, 'wikipedia_races.json')

# Save Tour de France stage data
save_to_csv(tour_stages, 'tour_stages.csv')
save_to_json(tour_stages, 'tour_stages.json')

print("\n✓ All data saved successfully!")
```
